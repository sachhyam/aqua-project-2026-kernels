```python
import numpy as np
from scipy.optimize import minimize


class MMRSolver:

    # =====================================================
    # init
    # =====================================================

    def __init__(self, x_train, kernel):

        self.x_train = x_train
        self.Y = x_train.copy()

        # user-supplied kernel function
        self.kernel = kernel

        # precompute kernel matrices
        self.K = self.build_kernel_matrix()

    # =====================================================
    # kernel matrices
    # =====================================================

    def build_kernel_matrix(self):

        return np.array([
            [self.kernel(xi, yj) for yj in self.Y]
            for xi in self.x_train
        ])

    def build_derivative_matrix(self, dkernel_dx):

        return np.array([
            [dkernel_dx(xi, yj) for yj in self.Y]
            for xi in self.x_train
        ])

    def build_second_derivative_matrix(self, d2kernel_dx2):

        return np.array([
            [d2kernel_dx2(xi, yj) for yj in self.Y]
            for xi in self.x_train
        ])

    # =====================================================
    # model
    # =====================================================

    def model(self, alpha, b, K=None):

        if K is None:
            K = self.K

        return b + K @ alpha

    # =====================================================
    # optimization
    # =====================================================

    def solve(self, loss_function, method="BFGS"):

        init = np.zeros(len(self.Y) + 1)

        self.loss_history = []

        def wrapped_loss(params):

            value = loss_function(params)

            self.loss_history.append(value)

            return value

        result = minimize(
            wrapped_loss,
            init,
            method=method
        )

        self.alpha_opt = result.x[:-1]
        self.b_opt = result.x[-1]

        return result

    # =====================================================
    # prediction
    # =====================================================

    def predict(self, x_test):

        K_test = np.array([
            [self.kernel(xi, yj) for yj in self.Y]
            for xi in x_test
        ])

        return self.b_opt + K_test @ self.alpha_opt
```

```python
import numpy as np
import matplotlib.pyplot as plt

# =====================================================
# kernel
# =====================================================

sigma = 0.15

def kernel(x, y):
    return np.exp(
        -(x - y)**2 / (2 * sigma**2)
    )

def dkernel_dx(x, y):
    return (
        -(x - y)/(sigma**2)
    ) * kernel(x, y)

def d2kernel_dx2(x, y):
    return (
        ((x - y)**2 - sigma**2)
        / sigma**4
    ) * kernel(x, y)

# =====================================================
# training points
# =====================================================

x_train = np.linspace(0, 1, 40)

# =====================================================
# create solver
# =====================================================

solver = MMRSolver(
    x_train=x_train,
    kernel=kernel
)

# derivative matrix
K_dx = solver.build_derivative_matrix(
    dkernel_dx
)

K_d2x = solver.build_second_derivative_matrix(
    d2kernel_dx2
)

# =====================================================
# oscillatory linear DE
#
# f'(x)
# =
# -2f - exp(-2x)sin(20x)
#
# exact:
#
# f(x)=exp(-2x)cos(20x)
# =====================================================

def g(x, f, f_dx):
    return -4*f

# =====================================================
# loss function
# =====================================================

def loss(params):

    alpha = params[:-1]
    b = params[-1]

    # function
    f = solver.model(alpha, b)

    # first derivative
    f_dx = K_dx @ alpha

    # second derivative
    f_d2x = K_d2x @ alpha

    # DE residual
    residual = f_d2x - g(x_train, f, f_dx)

    de_loss = np.mean(
        residual**2
    )

    # boundary conditions
    bc1 = (f[0] - 1.0)**2
    bc2 = (f_dx[0] - 0.0)**2

    return de_loss + 10*(bc1 + bc2)

# =====================================================
# solve
# =====================================================

result = solver.solve(loss, method="L-BFGS-B")

print("success:", result.success)

# =====================================================
# prediction
# =====================================================

x_test = np.linspace(0, 1, 400)

y_pred = solver.predict(x_test)

# exact solution
y_exact = y_exact = np.cos(2*x_test)

# =====================================================
# plot
# =====================================================

plt.figure(figsize=(10,5))

plt.plot(
    x_test,
    y_exact,
    label="exact"
)

plt.plot(
    x_test,
    y_pred,
    "--",
    label="MMR"
)

plt.legend()
plt.grid()
plt.show()
```
