```python
import numpy as np
from scipy.optimize import minimize


class MMRSolver:

    # =====================================================
    # init
    # =====================================================

    def __init__(self, x_train, kernel):

        self.x_train = x_train
        self.Y = x_train.copy()

        # user-supplied kernel function
        self.kernel = kernel

        # precompute kernel matrices
        self.K = self.build_kernel_matrix()

    # =====================================================
    # kernel matrices
    # =====================================================

    def build_kernel_matrix(self):

        return np.array([
            [self.kernel(xi, yj) for yj in self.Y]
            for xi in self.x_train
        ])

    def build_derivative_matrix(self, dkernel_dx):

        return np.array([
            [dkernel_dx(xi, yj) for yj in self.Y]
            for xi in self.x_train
        ])

    def build_second_derivative_matrix(self, d2kernel_dx2):

        return np.array([
            [d2kernel_dx2(xi, yj) for yj in self.Y]
            for xi in self.x_train
        ])

    # =====================================================
    # model
    # =====================================================

    def model(self, alpha, b, K=None):

        if K is None:
            K = self.K

        return b + K @ alpha

    # =====================================================
    # optimization
    # =====================================================

    def solve(self, loss_function, method="BFGS"):

        init = np.zeros(len(self.Y) + 1)

        self.loss_history = []

        def wrapped_loss(params):

            value = loss_function(params)

            self.loss_history.append(value)

            return value

        result = minimize(
            wrapped_loss,
            init,
            method=method
        )

        self.alpha_opt = result.x[:-1]
        self.b_opt = result.x[-1]

        return result

    # =====================================================
    # prediction
    # =====================================================

    def predict(self, x_test):

        K_test = np.array([
            [self.kernel(xi, yj) for yj in self.Y]
            for xi in x_test
        ])

        return self.b_opt + K_test @ self.alpha_opt
```

```python
import numpy as np

sigma = 0.2

def rbf_kernel(x, y):
    return np.exp(-(x - y)**2 / (2 * sigma**2))

def drbf_kernel_dx(x, y):
    return -(x - y) / sigma**2 * rbf_kernel(x, y)



x_train = np.linspace(0, 5, 50)
x_test = np.linspace(0, 5, 25)

```

```python


exact = 0.5 * (
    np.sin(x_test)
    - np.cos(x_test)
    + np.exp(-x_test)
)
```

```python
import numpy as np

from qiskit import QuantumCircuit, transpile
from qiskit.quantum_info import Statevector
from functools import partial
from qiskit_aer import AerSimulator
from qiskit_ibm_runtime import QiskitRuntimeService

# service = QiskitRuntimeService()

# hardware_backend = service.least_busy(simulator=False, operational=True)
```

```python
# import numpy as np

# from qiskit import QuantumCircuit, transpile
# from qiskit.quantum_info import Statevector

N_QUBITS = 8

def feature_map(x):

    qc = QuantumCircuit(N_QUBITS)

    for q in range(N_QUBITS):
        qc.rx((q + 1) * x, q)

    return qc

def kernel_circuit(x, y):

    qc = QuantumCircuit(N_QUBITS)

    qc.compose(
        feature_map(y),
        inplace=True
    )

    qc.compose(
        feature_map(x).inverse(),
        inplace=True
    )

    return qc


# def quantum_kernel(x, y):

#     qc = QuantumCircuit(N_QUBITS)

#     qc.compose(
#         feature_map(y),
#         inplace=True
#     )

#     qc.compose(
#         feature_map(x).inverse(),
#         inplace=True
#     )

#     psi = Statevector.from_instruction(qc)

#     amp0 = psi.data[0]

#     return np.abs(amp0)**2

# def dquantum_kernel_dx(
#     x,
#     y,
#     eps=1e-5
# ):

#     return (
#         quantum_kernel(x + eps, y)
#         -
#         quantum_kernel(x - eps, y)
#     ) / (2 * eps)

def quantum_kernel(
    x,
    y,
    backend=None,
    shots=4096
):
    """
    If backend=None:
        exact statevector simulation

    Otherwise:
        run on simulator or hardware backend
        using shot estimation.
    """

    qc = kernel_circuit(x, y)

    # -----------------------------------
    # exact simulation
    # -----------------------------------

    if backend is None:

        psi = Statevector.from_instruction(qc)

        return np.abs(psi.data[0])**2

    # -----------------------------------
    # hardware / shot-based simulation
    # -----------------------------------

    qc = qc.copy()
    qc.measure_all()

    tqc = transpile(
        qc,
        backend,
        optimization_level=1
    )

    job = backend.run(
        tqc,
        shots=shots
    )

    counts = job.result().get_counts()

    zero_state = "0" * N_QUBITS

    return counts.get(zero_state, 0) / shots

def dquantum_kernel_dx(
    x,
    y,
    backend=None,
    shots=4096,
    eps=1e-3
):

    return (
        quantum_kernel(
            x + eps,
            y,
            backend=backend,
            shots=shots
        )
        -
        quantum_kernel(
            x - eps,
            y,
            backend=backend,
            shots=shots
        )
    ) / (2 * eps)

```

```python
def train_solver(kernel, dkernel_dx, x_train):

    solver = MMRSolver(
        x_train=x_train,
        kernel=kernel
    )

    K = solver.K
    Kp = solver.build_derivative_matrix(dkernel_dx)

    def loss(params):

        alpha = params[:-1]
        b = params[-1]

        f = b + K @ alpha
        df = Kp @ alpha

        residual = df + f - np.sin(x_train)

        return (
            np.mean(residual**2)
            + f[0]**2
        )

    solver.solve(loss)

    return solver
```

```python
shots = 20000

aer_backend = AerSimulator()

qk_exact = quantum_kernel

dqk_exact = dquantum_kernel_dx

qk_shots = partial(
    quantum_kernel,
    backend=aer_backend,
    shots=shots
)

dqk_shots = partial(
    dquantum_kernel_dx,
    backend=aer_backend,
    shots=shots
)

# qk_hw = partial(
#     quantum_kernel,
#     backend=hardware_backend,
#     shots=8192
# )

# dqk_hw = partial(
#     dquantum_kernel_dx,
#     backend=hardware_backend,
#     shots=8192
# )


```

```python
rbf_solver = train_solver(
    kernel=rbf_kernel,
    dkernel_dx=drbf_kernel_dx,
    x_train=x_train
)

```

```python
qk_exact_solver = train_solver(
    kernel=qk_exact,
    dkernel_dx=dqk_exact,
    x_train=x_train
)

```

```python
qk_shot_solver = train_solver(
    kernel=qk_shots,
    dkernel_dx=dqk_shots,
    x_train=x_train
)

# qk_hw_solver = train_solver(
#     kernel=qk_hw,
#     dkernel_dx=dqk_hw,
#     x_train=x_train
# )
```

```python
rbf_pred = rbf_solver.predict(x_test)

qk_exact_pred = qk_exact_solver.predict(x_test)

qk_shot_pred = qk_shot_solver.predict(x_test)

# qk_hw_pred = qk_hw_solver.predict(x_test)
```
